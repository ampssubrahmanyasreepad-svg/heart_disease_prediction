```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import joblib

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

print("Libraries imported successfully.")
```

```python
df = pd.read_csv("C:\\Users\\User\\Downloads\\heart+disease\\heart_disease_combined_raw(3).csv")

print("Dataset Shape:", df.shape)

print("\nDataset Preview:")
display(df.head())
```

```python
df_model = df.copy()

# Convert original target into binary classification
df_model["target"] = (df_model["num"] > 0).astype(int)

print("Original target values:")
print(df_model["num"].value_counts().sort_index())

print("\nBinary target distribution:")
print(df_model["target"].value_counts())

print("\nTarget meaning:")
print("0 = No Heart Disease")
print("1 = Heart Disease Present")
```

```python
features = [
    "age",
    "sex",
    "cp",
    "trestbps",
    "chol",
    "fbs",
    "restecg",
    "thalach",
    "exang",
    "oldpeak",
    "slope",
    "ca",
    "thal"
]

X = df_model[features].copy()
y = df_model["target"].copy()

# Convert possible non-numeric values to NaN
X = X.replace(["?", "NA", "N/A"], np.nan)

for column in X.columns:
    X[column] = pd.to_numeric(X[column], errors="coerce")

print("Final feature count:", X.shape[1])
print("Final dataset shape:", X.shape)

print("\nSelected features:")
for i, feature in enumerate(features, 1):
    print(i, "-", feature)
```

```python
print("Missing values before imputation:")
display(X.isnull().sum())

# Median imputation
for column in X.columns:
    X[column] = X[column].fillna(X[column].median())

print("\nTotal missing values after imputation:")
print(X.isnull().sum().sum())
```

```python
# Load the model selected by Notebook 09
selected_model = joblib.load("final_heart_disease_model.pkl")

# Read the selected model name
with open("best_model_name.txt", "r") as file:
    best_model_name = file.read().strip()

print("=" * 55)
print("FINAL MODEL")
print("=" * 55)

print("Selected Model:", best_model_name)

print("\nModel Configuration:")
print(selected_model)
```

```python
# Retrain selected model on the complete prepared dataset
final_model = selected_model

final_model.fit(X, y)

print("Final model trained successfully.")
print("Training records used:", len(X))
print("Features used:", X.shape[1])
```
